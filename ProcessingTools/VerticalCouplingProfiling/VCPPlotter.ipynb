# VCPPlotter

**VCPPlotter** (*Vertical Coupling Profiling Plotter*) is a data visualization script developed to process and visualize measurements obtained from vertical coupling profiling experiments.

Step 1: Import the necessary libraries.

```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Times New Roman"
plt.rcParams["font.size"] = 14
```

Step 2: Load and group the measurement data

```python
filepath = Path("vertical-2026-07-31_15-16-17.csv") 
# filepath = Path("vertical-2026-07-30_15-55-10.csv") 
df = pd.read_csv(filepath)
# Group measurements by waveguide
df.groupby("Waveguide")
```

Step 3: Inspect the grouped data

```python
# Inspect the contents of each group
for nombre, grupo in df.groupby("Waveguide"):
    print(nombre)
    print(grupo)
    print("-" * 40)
```

Step 5: Generate and save the power histogram for each of the optical structures

## WG0 - Waveguide

```python
# Select WG0
potencia = df[df["Waveguide"] == "WG0"]["Power_dBm"]

# Statistics
media = potencia.mean()
desv_std = potencia.std()

print(f"Mean: {media:.3f} dBm")
print(f"Standard deviation: {desv_std:.3f} dB")

# Generate histogram
# plt.figure(figsize=(8, 8))
plt.figure(figsize=(10, 5))

plt.hist(
    potencia,
    bins="auto",
    color="#333745",
    edgecolor="black",
)

plt.title("Power Histogram - WG0")
plt.xlabel("Power (dBm)")
plt.ylabel("Frequency")

plt.grid(alpha=0.3)

plt.savefig("WG0.png", bbox_inches="tight")
plt.show()
```
